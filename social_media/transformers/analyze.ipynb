# processing

```python
import pandas as pd
import spacy
```

```python
# loading our language model and data

nlp = spacy.load("en_core_web_sm")
df = pd.read_json('cleaned.json', lines=True)
```

```python
texts = df['text'].to_list()
```

```python
texts[0]
```

```python
# doing just the first article text, because otherwise it's too much data
dataset = ''.join(texts[:100])
```

```python
type(dataset)
```

```python
len(dataset)
```

```python
# processing the text column into an NLP "doc"

docs = nlp(dataset)
```

```python
len(docs)
```

```python
## let's check out what kind of object this is, what can we do with it?

dir(docs)
```

```python
for sent in docs[:1000].sents:
    print(sent)
```

```python
with open('output.txt', 'w') as f:
    for sent in docs.sents:
        f.write(str(sent) + '\n')
```

```python
with open('output.txt', 'r') as f:
    sents = f.readlines()
```

```python
# remove sents that start with >>>

for i in sents:
    if i.startswith('>>>'):
        sents.remove(i)
```

```python
# write sents to text file
with open('output_cleaned.txt', 'w') as f:
    for sent in sents:
        f.write(sent)
```
